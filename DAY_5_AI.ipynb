```python
#IMPORT LIBRARIES
import sklearn
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
#Create Dataset
sentences = [
    "Artificial Intelligence is transforming industries.",
    "Machine learning helps systems learn from data.",
    "Deep learning is a subset of machine learning.",
    "Neural networks are inspired by the human brain.",
    "AI is used in healthcare for diagnosis.",
    "Chatbots improve customer service experience.",
    "Natural Language Processing enables machines to understand text.",
    "Computer vision helps machines see images.",
    "Self-driving cars use AI technologies.",
    "AI can automate repetitive tasks.",
    "Data is the fuel for machine learning models.",
    "Big data helps improve AI accuracy.",
    "AI is used in recommendation systems.",
    "Speech recognition converts voice to text.",
    "AI can detect fraud in banking.",
    "Robotics uses AI for automation.",
    "AI is widely used in education technology.",
    "Machine learning models require training data.",
    "AI ethics is an important topic today.",
    "AI is shaping the future of technology."
]

#Load model and embedding
model = SentenceTransformer('all-MiniLM-L6-v2')

# Convert sentences into vectors
embeddings = model.encode(sentences)


# Semantic Search Function
def semantic_search(query, top_k=3):
    query_embedding = model.encode([query])
    similarities = cosine_similarity(query_embedding, embeddings)[0]

    top_indices = np.argsort(similarities)[-top_k:][::-1]

    results = [(sentences[i], similarities[i]) for i in top_indices]
    return results

# Keyword Search Function

def keyword_search(query, top_k=3):
    query_words = set(query.lower().split())

    scores = []
    for sentence in sentences:
        sentence_words = set(sentence.lower().split())
        score = len(query_words.intersection(sentence_words))
        scores.append(score)

    top_indices = np.argsort(scores)[-top_k:][::-1]

    results = [(sentences[i], scores[i]) for i in top_indices]
    return results

# Test Queries

queries = [
    "AI in healthcare",
    "machines understanding text",
    "future of AI technology"
]
# Run & Compare Results
for query in queries:
    print("\n🔍 Query:", query)

    print("\n👉 Semantic Search Results:")
    for res in semantic_search(query):
        print(f"- {res[0]} (score: {res[1]:.4f})")

    print("\n👉 Keyword Search Results:")
    for res in keyword_search(query):
        print(f"- {res[0]} (score: {res[1]})")

    print("\n" + "="*50)
```
