```python
!pip install nltk

import nltk
import re
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# 3 noisy real-world text samples

social_media = "OMG!!! 😂 I just loooove this phone!!! #awesome #AI 🤖 It's sooo coool!!!"
news_article = "Breaking News: The Government, on Monday, announced new policies!!! Experts are unsure... 🤔"
customer_review = "I bought this product last week. It's okay, not great!!! 😐 Could be better..."

samples = {
    "Social Media": social_media,
    "News Article": news_article,
    "Customer Review": customer_review
}
```

```python
import nltk
nltk.download('punkt_tab')

print("🔹 TOKENIZATION OUTPUT\n")

for name, text in samples.items():
    print(f"\n{name}:\n")

    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    print("Sentences:", sentences)
    print("Words:", words)
```

```python
stop_words = set(stopwords.words('english'))

print("\n🔹 STOPWORD REMOVAL\n")

for name, text in samples.items():
    words = word_tokenize(text.lower())

    before_vocab = set(words)
    filtered_words = [w for w in words if w not in stop_words]
    after_vocab = set(filtered_words)

    print(f"\n{name}:")
    print("Before vocab size:", len(before_vocab))
    print("After vocab size:", len(after_vocab))
    print("Filtered words:", filtered_words)
```

```python
print("\n🔹 REGEX CLEANING (REMOVE PUNCTUATION)\n")

for name, text in samples.items():

    # Removing punctuation is important because punctuation adds noise
    # and does not contribute meaningful semantic value in most NLP tasks.
    clean_text = re.sub(r'[^\w\s]', '', text)

    print(f"\n{name}:")
    print("Before:", text)
    print("After:", clean_text)
```

```python
ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("\n🔹 STEMMING vs LEMMATIZATION\n")

for name, text in samples.items():
    words = word_tokenize(text.lower())

    stemmed = [ps.stem(w) for w in words]
    lemmatized = [lemmatizer.lemmatize(w) for w in words]

    print(f"\n{name}:")
    print("Original:", words)
    print("Stemmed:", stemmed)
    print("Lemmatized:", lemmatized)
```

```python
print("\n🔹 FULL PIPELINE COMPARISON\n")

for name, text in samples.items():

    # Lowercase
    lower = text.lower()

    # Remove punctuation
    no_punct = re.sub(r'[^\w\s]', '', lower)

    # Tokenize
    tokens = word_tokenize(no_punct)

    # Remove stopwords
    filtered = [w for w in tokens if w not in stop_words]

    print(f"\n{name}:")
    print("Original:", text)
    print("Processed:", filtered)
```
